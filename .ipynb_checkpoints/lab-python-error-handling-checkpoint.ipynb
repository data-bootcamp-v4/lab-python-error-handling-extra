Objective: Practice how to identify, handle and recover from potential errors in Python code using try-except blocks.

## Challenge 

Paste here your lab *functions* solutions. Apply error handling techniques to each function using try-except blocks. 

The try-except block in Python is designed to handle exceptions and provide a fallback mechanism when code encounters errors. By enclosing the code that could potentially throw errors in a try block, followed by specific or general exception handling in the except block, we can gracefully recover from errors and continue program execution.

However, there may be cases where an input may not produce an immediate error, but still needs to be addressed. In such situations, it can be useful to explicitly raise an error using the "raise" keyword, either to draw attention to the issue or handle it elsewhere in the program.

Modify the code to handle possible errors in Python, it is recommended to use `try-except-else-finally` blocks, incorporate the `raise` keyword where necessary, and print meaningful error messages to alert users of any issues that may occur during program execution.



```python
def initialize_inventory(products):
    inventory = {}

    try:
        if not isinstance(products, (list, tuple, set)):
            raise TypeError("Products must be a list, tuple, or set.")

        if not products:
            raise ValueError("The product list cannot be empty.")

        for product in products:
            if not isinstance(product, str) or not product.strip():
                raise ValueError("Product names must be non-empty strings.")

            quantity = int(input(f"Enter quantity for {product}: "))

            if quantity < 0:
                raise ValueError("Quantity cannot be negative.")

            inventory[product] = quantity

    except (TypeError, ValueError) as error:
        print(f"Inventory initialization error: {error}")
        return None

    else:
        print("Inventory initialized successfully.")
        return inventory

    finally:
        print("Inventory initialization finished.")


def get_customer_orders():
    customer_orders = set()

    try:
        while True:
            product = input(
                "Enter product name (or 'done' to finish): "
            ).strip()

            if product.lower() == "done":
                break

            if not product:
                print("Error: Product name cannot be empty.")
                continue

            customer_orders.add(product)

    except (EOFError, KeyboardInterrupt) as error:
        print(f"Order input interrupted: {error}")

    else:
        print("Customer orders collected successfully.")
        return customer_orders

    finally:
        print("Customer order collection finished.")

    return customer_orders


def update_inventory(customer_orders, inventory):
    try:
        if not isinstance(customer_orders, set):
            raise TypeError("Customer orders must be a set.")

        if not isinstance(inventory, dict):
            raise TypeError("Inventory must be a dictionary.")

        for product in customer_orders:
            if product not in inventory:
                print(f"Warning: '{product}' is not in inventory.")
                continue

            if inventory[product] <= 0:
                print(f"Warning: '{product}' is out of stock.")
                continue

            inventory[product] -= 1

    except (TypeError, ValueError, KeyError) as error:
        print(f"Inventory update error: {error}")
        return None

    else:
        print("Inventory updated successfully.")
        return inventory

    finally:
        print("Inventory update finished.")


def calculate_order_statistics(customer_orders, products):
    try:
        if not isinstance(customer_orders, set):
            raise TypeError("Customer orders must be a set.")

        if not isinstance(products, (list, tuple, set)):
            raise TypeError("Products must be a list, tuple, or set.")

        if not products:
            raise ValueError("Cannot calculate percentage with no products.")

        if not customer_orders.issubset(set(products)):
            raise ValueError("Orders contain products outside the product list.")

        total_products_ordered = len(customer_orders)
        percentage_ordered = (
            total_products_ordered / len(set(products))
        ) * 100

    except (TypeError, ValueError, ZeroDivisionError) as error:
        print(f"Statistics calculation error: {error}")
        return None

    else:
        print("Order statistics calculated successfully.")
        return total_products_ordered, percentage_ordered

    finally:
        print("Statistics calculation finished.")


def print_order_statistics(order_statistics):
    try:
        if order_statistics is None:
            raise ValueError("No order statistics are available.")

        total, percentage = order_statistics

        print(f"Total unique products ordered: {total}")
        print(f"Percentage of unique products ordered: {percentage:.2f}%")

    except (TypeError, ValueError) as error:
        print(f"Error displaying order statistics: {error}")

    else:
        print("Order statistics displayed successfully.")

    finally:
        print("Statistics display finished.")


def print_updated_inventory(inventory):
    try:
        if not isinstance(inventory, dict):
            raise TypeError("Inventory must be a dictionary.")

        print("\nUpdated Inventory:")

        for product, quantity in inventory.items():
            print(f"{product}: {quantity}")

    except (TypeError, AttributeError) as error:
        print(f"Error displaying inventory: {error}")

    else:
        print("Updated inventory displayed successfully.")

    finally:
        print("Inventory display finished.")


# Main program
if __name__ == "__main__":
    products = ["apple", "banana", "orange"]

    inventory = initialize_inventory(products)

    if inventory is not None:
        customer_orders = get_customer_orders()

        updated_inventory = update_inventory(
            customer_orders, inventory
        )

        order_statistics = calculate_order_statistics(
            customer_orders, products
        )

        print_order_statistics(order_statistics)

        if updated_inventory is not None:
            print_updated_inventory(updated_inventory)
```
