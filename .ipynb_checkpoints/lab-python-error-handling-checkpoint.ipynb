## Challenge 

Paste here your lab *functions* solutions. Apply error handling techniques to each function using try-except blocks. 

The try-except block in Python is designed to handle exceptions and provide a fallback mechanism when code encounters errors. By enclosing the code that could potentially throw errors in a try block, followed by specific or general exception handling in the except block, we can gracefully recover from errors and continue program execution.

However, there may be cases where an input may not produce an immediate error, but still needs to be addressed. In such situations, it can be useful to explicitly raise an error using the "raise" keyword, either to draw attention to the issue or handle it elsewhere in the program.

Modify the code to handle possible errors in Python, it is recommended to use `try-except-else-finally` blocks, incorporate the `raise` keyword where necessary, and print meaningful error messages to alert users of any issues that may occur during program execution.



```python
# Lab | Error Handling


products = ["t-shirt", "mug", "hat", "book", "keychain"]


# 1. Initialize inventory
def initialize_inventory(products):
    inventory = {}

    for product in products:
        while True:
            try:
                quantity = int(
                    input(f"Enter the quantity of {product}s available: ")
                )

                if quantity < 0:
                    raise ValueError(
                        "Quantity cannot be negative."
                    )

            except ValueError as error:
                print("Error:", error)

            else:
                inventory[product] = quantity
                break

            finally:
                print(f"Finished checking quantity for {product}.")

    return inventory


# 2. Get customer orders
def get_customer_orders(inventory):

    while True:
        try:
            number_of_orders = int(
                input("Enter the number of customer orders: ")
            )

            if number_of_orders < 0:
                raise ValueError(
                    "Number of orders cannot be negative."
                )

        except ValueError as error:
            print("Error:", error)

        else:
            break

        finally:
            print("Finished checking number of orders.")

    customer_orders = set()

    for _ in range(number_of_orders):

        while True:
            try:
                product = input(
                    "Enter the name of a product: "
                ).lower()

                if product not in inventory:
                    raise ValueError(
                        "Product does not exist in inventory."
                    )

                if inventory[product] <= 0:
                    raise ValueError(
                        "Product is out of stock."
                    )

            except ValueError as error:
                print("Error:", error)

            else:
                customer_orders.add(product)
                break

            finally:
                print("Finished checking product.")

    return customer_orders


# 3. Update inventory
def update_inventory(customer_orders, inventory):

    try:
        updated_inventory = {
            product: quantity - 1
            if product in customer_orders
            else quantity
            for product, quantity in inventory.items()
        }

        updated_inventory = {
            product: quantity
            for product, quantity in updated_inventory.items()
            if quantity > 0
        }

    except Exception as error:
        print("Error while updating inventory:", error)
        return inventory

    else:
        print("Inventory updated successfully.")
        return updated_inventory

    finally:
        print("Finished inventory update process.")


# 4. Calculate order statistics
def calculate_order_statistics(customer_orders, products):

    try:
        if len(products) == 0:
            raise ValueError(
                "Product list cannot be empty."
            )

        total_products_ordered = len(customer_orders)

        percentage_unique_products = (
            total_products_ordered / len(products)
        ) * 100

    except ValueError as error:
        print("Error:", error)
        return 0, 0

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
        return 0, 0

    else:
        return (
            total_products_ordered,
            percentage_unique_products
        )

    finally:
        print("Finished calculating order statistics.")


# 5. Print order statistics
def print_order_statistics(order_statistics):

    try:
        total_products_ordered = order_statistics[0]
        percentage_unique_products = order_statistics[1]

        print("\nOrder Statistics:")
        print(
            "Total Products Ordered:",
            total_products_ordered
        )

        print(
            "Percentage of Unique Products Ordered:",
            percentage_unique_products
        )

    except IndexError:
        print(
            "Error: Order statistics are incomplete."
        )

    except TypeError:
        print(
            "Error: Invalid order statistics format."
        )

    finally:
        print("Finished printing order statistics.")


# 6. Print updated inventory
def print_updated_inventory(inventory):

    try:
        print("\nUpdated Inventory:")

        if not inventory:
            raise ValueError(
                "Inventory is empty."
            )

        for product, quantity in inventory.items():
            print(f"{product}: {quantity}")

    except ValueError as error:
        print("Error:", error)

    except AttributeError:
        print(
            "Error: Inventory must be a dictionary."
        )

    finally:
        print("Finished printing inventory.")


# 7. Calculate total price
def calculate_total_price(customer_orders):

    prices = {}

    for product in customer_orders:

        while True:
            try:
                price = float(
                    input(
                        f"Enter the price of {product}: "
                    )
                )

                if price < 0:
                    raise ValueError(
                        "Price cannot be negative."
                    )

            except ValueError as error:
                print("Error:", error)

            else:
                prices[product] = price
                break

            finally:
                print(
                    f"Finished checking price for {product}."
                )

    try:
        total_price = sum(
            price
            for price in prices.values()
        )

    except TypeError as error:
        print(
            "Error calculating total price:",
            error
        )
        return 0

    else:
        return total_price

    finally:
        print("Finished calculating total price.")


# MAIN PROGRAM

try:

    inventory = initialize_inventory(products)

    customer_orders = get_customer_orders(
        inventory
    )

    order_statistics = calculate_order_statistics(
        customer_orders,
        products
    )

    print_order_statistics(
        order_statistics
    )

    inventory = update_inventory(
        customer_orders,
        inventory
    )

    print_updated_inventory(
        inventory
    )

    total_price = calculate_total_price(
        customer_orders
    )

except Exception as error:

    print(
        "Unexpected program error:",
        error
    )

else:

    print(
        "\nTotal Price:",
        total_price
    )

finally:

    print(
        "\nProgram finished."
    )
```
